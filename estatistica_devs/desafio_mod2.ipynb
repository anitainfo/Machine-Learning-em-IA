## Desafio Estatística para Devs - Módulo 2

```python
# Código Base
import pandas as pd
import matplotlib.pyplot as plt

# Dicionário de faturamento
dict_faturamento = {
    'data_ref': [
        '2023-01-01', 
        '2020-02-01', 
        '2021-03-01', 
        '2022-04-01', 
        '2023-05-01',
        '2023-06-01', 
        '2020-07-01', 
        '2021-08-01', 
        '2022-09-01', 
        '2023-10-01',
        '2022-11-01', 
        '2023-12-01',
        ],
    'valor': [
        400000, 
        890000, 
        760000, 
        430000, 
        920000,
        340000, 
        800000, 
        500000, 
        200000, 
        900000,
        570000, 
        995000,
        ]
}
```

```python
# Teste
dict_faturamento
```

### Criação do DataFrame

```python
df_faturamento = pd.DataFrame.from_dict(dict_faturamento)
```

```python
# Teste
df_faturamento
```

```python
# Correção de data_ref como datetime
df_faturamento["data_ref"] = pd.to_datetime(df_faturamento["data_ref"])
```

```python
# Teste
df_faturamento.info()
```

### Tarefas

```python
# 1. Trazer a média das vendas
media_vendas = df_faturamento.valor.mean()
media_vendas
```

```python
# 2. Criar um gráfico de barras vertical mostrando o mês 
# de referência e o valor
ax = df_faturamento.sort_values('data_ref').plot.bar(x='data_ref', y='valor')
ax.set_xticklabels([d.strftime('%m/%Y') for d in df_faturamento.sort_values('data_ref')['data_ref']])
ax.ticklabel_format(style='plain', axis='y')
```

```python
# 3. Criar um gráfico de linhas mostrando o mês de referência e o valor
ax = df_faturamento.sort_values('data_ref').plot.line(x='data_ref', y='valor')
ax.set_xticklabels([d.strftime('%m/%Y') for d in df_faturamento.sort_values('data_ref')['data_ref']])
ax.ticklabel_format(style='plain', axis='y')
```
